```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
def load_data(file_path):
    df=pd.read_csv(file_path)
    return df
```

```python
def basic_inf0(df):
    try:
        print("DataFrame shape:", df.shape)
        print("\nDataFrame info:")
        print(df.info())
        print("\nDataFrame description:")
        print(df.describe())
    except Exception as e:
        print("An error occurred while getting basic information:", e)
```

```python
def cleaning(df):
    try:
        if df.isnull().sum().sum()>0:
            print("Missing values found")
            df=df.dropna()
        else:
            print("No missing values found")
        if df.duplicated().sum()>0:
            print("Duplicate values found")
            df=df.drop_duplicates()
        else:
            print("No duplicate values found")
        return df
    except Exception as e:
        print("An error occurred during data cleaning:", e)
            
```

```python
def discription(df):
    try:
        print("DataFrame description:")
        print(df.describe())
    except Exception as e:
        print("An error occurred while describing the DataFrame:", e)
```

```python
def plot_histogram(df):
    try:
        numeric_cols = df.select_dtypes(include=[np.number]).columns
        for col in numeric_cols:
            plt.figure(figsize=(8, 6))
            plt.hist(df[col], bins=30, edgecolor='k')
            plt.title(f'Histogram of {col}')
            plt.xlabel(col)
            plt.ylabel('Frequency')
            plt.grid(axis='y', alpha=0.75)
            plt.show()
    except Exception as e:
        print("An error occurred while plotting histograms:", e)
```

```python
def plot_boxplot(df):
    try:
        numeric_cols = df.select_dtypes(include=[np.number]).columns
        for col in numeric_cols:
            plt.figure(figsize=(8, 6))
            plt.boxplot(df[col].dropna(), vert=False)
            plt.title(f'Boxplot of {col}')
            plt.xlabel(col)
            plt.grid(axis='x', alpha=0.75)
            plt.show()
    except Exception as e:
        print("An error occurred while plotting boxplots:", e)
```

```python
def main():
    file_path = '../../week4/task1/cleaned_students.csv'  # Update with your actual file path
    df = load_data(file_path)
    basic_inf0(df)
    df = cleaning(df)
    discription(df)
    plot_histogram(df)
    plot_boxplot(df)
    
if __name__ == "__main__":
    main()
```
