```python
import pandas as pd
import numpy as np
import plotly.express as px

df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv')
df = df.dropna(subset=['age', 'fare', 'embark_town'])
df['survived_label'] = df['survived'].map({0: 'Did not survive', 1: 'Survived'})
```

```python
summary = df.groupby(['class', 'sex'], as_index=False).agg(
    survival_rate=('survived', 'mean'),
    passenger_count=('survived', 'size')
)

fig1 = px.bar(
    summary, x='class', y='survival_rate', color='sex', barmode='group',
    hover_data={'passenger_count': True, 'survival_rate': ':.0%'},
    title='Survival Rate by Passenger Class and Sex',
    labels={'survival_rate': 'Survival Rate', 'class': 'Passenger Class', 'sex': 'Sex'}
)
fig1.update_layout(yaxis_tickformat='.0%')
fig1.write_html('fig1_survival_rate_by_class.html')
fig1.show()
```

```python
fig2 = px.scatter(
    df, x='age', y='fare', color='survived_label',
    hover_data=['class', 'embark_town'],
    title='Passenger Age vs. Fare, Coloured by Survival',
    labels={'age': 'Age (years)', 'fare': 'Fare (GBP)', 'survived_label': 'Outcome'}
)
fig2.write_html('fig2_age_vs_fare.html')
fig2.show()
```

```python
fig3 = px.box(
    df, x='class', y='fare', color='survived_label',
    points='outliers',
    title='Fare Distribution by Passenger Class and Survival',
    labels={'fare': 'Fare (GBP)', 'class': 'Passenger Class', 'survived_label': 'Outcome'}
)
fig3.write_html('fig3_fare_by_class.html')
fig3.show()
```

```python
fig4 = px.sunburst(
    df, path=['class', 'sex', 'survived_label'],
    title='Passenger Composition by Class, Sex and Survival Outcome'
)
fig4.update_traces(textinfo='label+percent parent')
fig4.write_html('fig4_sunburst_composition.html')
fig4.show()
```
